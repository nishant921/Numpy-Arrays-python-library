RESHAPING

```python
import numpy as np
```

```python
# reshape
a = np.arange(12)
a.reshape(3,4)
```

```python
# transpose 
a = a.reshape(3,4)
a
# 1
np.transpose(a)
# 2
a.T

a2 = np.arange(10)
np.transpose(a2) # -> same as before
```

.ravel(): Convert any dimensions of array into one dimensional array

```python
r = np.arange(27).reshape(3,3,3)
print(r)

r.ravel()
```
