ARRAY OPERATIONS

```python
import numpy as np
```

```python
a1 = np.arange(12).reshape(3,4)
a2 = np.arange(12,24).reshape(3,4) 
```

```python
print(a1)
print(a2)
```

Operations : Scalar oprations And Vector Operations

```python
# Scalar operations

# 1. Arthemtic 
a1*2 # like matrix scalar multiplication
a1**2
a1/2
a1//2
a1+2
a1-2
```

```python
# Relational operator(>,<,>=,<=,!=,==)
a1>6
a2>=19
a2<10
a2==21
```

VECTOR OPERATIONS :

```python
# Arthemtic operations: Shape of the both array should be same  
print(a1)
print()
print(a2)

#  matrix addition
a1+a2
#  matrix subrtactiom
a1-a2

# element wise multiplication
a1*a2


# Relational: 
a1<a2
a1>=a2
a1==a2
```
