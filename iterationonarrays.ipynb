ITERATING ON ARRAYS

```python
import numpy as np
```

```python
# Using Loop
a1 = np.arange(10)
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(27).reshape(3,3,3)
```

```python
a1
for i in a1:
    print(i)
```

```python
a2
for i in a2:
    print(i)
```

```python
a3
for i in a3:
    print(i)
```

.nditer(<arryaname>): for iterating on every elements

```python
for i in np.nditer(a3):
    print(i)
```
