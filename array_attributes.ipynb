ARRAY ATTRIBUTES

```python
import numpy as np
```

```python
a1 = np.arange(10)
a2 = np.arange(12,dtype=float).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2)
```

```python
print(a1)
print()
print(a2)
print()
print(a3)
```

<!-- ndim: No. of n dimensional array  -->

```python
a1.ndim
a2.ndim
a3.ndim
```

<!-- shape:no. of rows and columns/no. of items -->

```python
a1.shape
a2.shape
a3.shape
```

<!-- size :total no of items in arrays -->

```python
a1.size
a2.size
a3.size
```

<!-- itemsize : size of item in the memory  -->

```python
a1.itemsize
a2.itemsize
a3.itemsize
a1 = np.arange(10,dtype=np.int32)
a1.itemsize
```

<!-- dtype : type of array -->

```python

a2.dtype
a3.dtype
a1.dtype
```

<!-- astype : to change the data type of an exisitng array -->

```python
a3.dtype
a3.astype(np.int32)
```
