STACKING

```python
import numpy as np
```

Condition : Array should be of same size 

```python
# Stacking: stacking up/joining two or more arrays into one -->horizontal -->vertical

a1 = np.arange(12).reshape(3,4)
a2 = np.arange(12,24).reshape(3,4)
a1
a2
```

```python
# 1. horizontal : .hstack(())
np.hstack((a1,a2))
```

```python
np.hstack((a1,a2,a1))
```

```python
# 2. vertical : .vstack(())
np.vstack((a1,a2))
```

```python
np.vstack((a1,a2,a2))
```

SPLITING : reverse functionality of stacking

```python
a1 = np.arange(12).reshape(3,4)
a2 = np.arange(12,24).reshape(3,4)

# horizontal split
np.hsplit(a1,2)
np.hsplit(a2,4)
```

```python
# Vertical split
np.vsplit(a2,3)
np.vsplit(a1,3)
```
