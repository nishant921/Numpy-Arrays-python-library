ARRAY FUNCTIONS 

```python
import numpy as np
```

```python
a1 = np.random.random((3,3))
a1 = np.round(a1*100)
a1
```

max/min/sum/prod

```python
np.max(a1)
np.min(a1)
np.sum(a1)
np.prod(a1)
# For every row or column find max/min/sum/prod
# axis = 0 -> y-axis/ col
# axis = 1 -> x-axis/ row
np.max(a1,axis=1)
np.min(a1,axis=0)

np.prod(a1,axis=1)
```

mean/median/std/var


```python
np.mean(a1)
np.mean(a1,axis=1)

np.median(a1)
np.median(a1,axis=1)


np.std(a1)
np.std(a1,axis=1)

np.var(a1)
np.var(a1,axis=1)
```

TRIGONOMETRIC FUNCTIONS: sim/cos/tan

```python
np.cos(a1)
np.sin(a1)
np.tan(a1)
```

dot product: no. of col = no. of row
-->shape: row of 1st * col of 2nd

```python
a2 =  np.arange(12).reshape(3,4)
a3 =  np.arange(12).reshape(4,3)

print(a2)
print()
print(a3)
```

```python
np.dot(a2,a3)
```

log and exponential 

```python
print(a1)
np.log(a1)
np.exp(a1)
```

round/floor/ceil

```python
r = np.random.random((3,4))*100
np.round(r)
np.floor(r)
np.ceil(r)
```
